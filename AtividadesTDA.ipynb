```python
'''
4. Faça um programa que solicita uma quantidade indefinida de notas de uma prova (deve parar quando um número negativo for digitado).
Armazena em uma lista.
Após isso, o programa deve mostrar a média das notas
'''
notas = []
while (nota := float(input("Digite uma nota (negativo para parar): "))) >= 0:
    notas.append(nota)
print(f"Média: {sum(notas)/len(notas):.2f}" if notas else "Nenhuma nota válida foi digitada")
```

```python
'''
5.Faça um programa que solicita uma quantidade indefinida de números positivos (deve parar quando um número negativo for digitado).
Armazene-os em duas listas:
uma para os números pares;
outra para os números ímpares.
A seguir, mostre a porcentagem de pares e ímpares digitados
'''
pares, impares = [], []
while (num := int(input("Digite um número (negativo para parar): "))) >= 0:
    (pares if num % 2 == 0 else impares).append(num)

total = len(pares) + len(impares)
if total > 0:
    print(f"Pares: {pares} - {len(pares)/total*100:.1f}%")
    print(f"Ímpares: {impares} - {len(impares)/total*100:.1f}%")
else:
    print("Nenhum número válido foi digitado.")
```

```python
'''
6. Faça um programa que solicita uma quantidade indefinida de números positivos (deve parar quando um número negativo for digitado).
Armazene todos os números digitados em uma lista, sem repetição
'''
numeros = set()
while (num := int(input("Digite um número (negativo para parar): "))) >= 0:
    numeros.add(num)
print(f"Lista sem repetição: {list(numeros)}")
```

```python
'''
7. Escreva um programa que solicita que o usuário digite a altura e o peso de 5 pessoas.
Verifique se pelo menos duas das pessoas tem a mesma altura e mesmo peso
'''
pessoas = []
for i in range(5):
    altura = float(input(f"Altura da pessoa {i+1} (m): "))
    peso = float(input(f"Peso da pessoa {i+1} (kg): "))
    pessoas.append((altura, peso))

if len(set(pessoas)) < 5:
    print("Há pelo menos duas pessoas com mesma altura e peso!")
else:
    print("Todas as pessoas têm altura e peso diferentes")
```

```python
'''
8. Faça um programa que solicita que o usuário digite o nome e a idade de diversas pessoas (o programa deve parar quando um nome vazio for informado).
Retorne o nome da pessoa mais velha
'''
mais_velha_nome, mais_velha_idade = "", -1
while (nome := input("Digite o nome (vazio para parar): ")):
    idade = int(input("Digite a idade: "))
    if idade > mais_velha_idade:
        mais_velha_nome, mais_velha_idade = nome, idade

if mais_velha_idade >= 0:
    print(f"Pessoa mais velha: {mais_velha_nome} com {mais_velha_idade} anos")
else:
    print("Nenhuma pessoa foi cadastrada.")
```
